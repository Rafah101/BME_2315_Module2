## Project Goal:
Facioscapulohumeral Muscular Dystrophy (FSHD) is one of the most common muscular dystrophies, inherited genetic disorders that involve muscle degradation. There is currently no treatment for the underlying mechanisms of FSHD, only treatments that help alleviate symptoms caused by it. Additionally, it is difficult to measure disease progression due to multiple factors, as explored below. Hence, the goal of this project is to determine which measure is the most effective for evaluating FSHD progression and should be included as an exploratory endpoint in a clinical trial during Phase 1/2 [1]. ## Disease Background: 
#### Prevalence & Incidence
Prevalence of FSHD is difficult to measure due to variation of this statistic in different studies and locations. For instance, one study that was conducted in the Netherlands found that the estimated prevalence rate was 12 per 100,000 individuals within the country, while the estimated incidence rate was 0.3 per 100,000 individuals within the country [2] . On the other hand, one study that was conducted in the Province of Padua, found that the estimated prevalence rate was 4.4 per 100,000 individuals within the Province of Padua, while there was no estimated incidence [3]. Determining the estimated prevalence and incidence of FSHD is difficult due to numerous factors. For instance, countless individuals with FSHD do not express severe symptoms of the disease and access to diagnostic equipment is dependent on location, as well as other socioeconomic factors [4]. Hence, the potential patient population is likely much larger than the reported estimated prevalence and incidence.

#### Risk Factors 
FSHD1, which accounts for approximately 95% of FSHD known cases, is caused by a contraction in the D4Z4 array on chromosome 4 at region 4q35 [7]. More specifically, there are 1 to 10 repeat units instead of the typical 11 to 100 [7]. This contraction leads to chromatin relaxation and hypomethylation. 

FSHD2, which accounts for approximately 5% of FSHD cases, is most commonly caused by a mutation in the SMCHD1 gene on chromosome 18 at region 18p11.32 [7]. In an individual without this mutation, the SMCHD1 gene produces a protein that binds to the D4Z4 array on chromosome 4 at region 4q35 and limits chromatin relaxation [7]. Contrastingly, if an individual does have this mutation, the SMCHD1 gene does not effectively produce this protein, prompting the chromatin to relax and hypomethylation to ensue [7].

In individuals with either form of FSHD, the chromatin relaxation and hypomethylation lead to the transcription of the DUX4 gene into mRNA, which is then translated into the DUX4 protein [7]. The DUX4 protein in skeletal muscle prompts apoptosis and additional disruptive pathways that lead to muscle cell degradation [7]. Mutations in two additional genes, DNMT3B and LRIF1, can also potentially lead to FSHD2 because they contribute to the prevention of the DUX4 gene transcription [8]. Importantly, FSHD1 and FSHD2 only arise if the D4Z4 array is located on a permissive 4qA allele because this allele includes a polyadenylation signal that prompts the stabilization of the DUX4 mRNA [9]. Hence, in individuals who do not have either form of FSHD and have typical methylation of the D4Z4 array, or whose array is located on a 4qB allele, DUX4 protein is not produced, and the muscle cells do not undergo degradation due to DUX4 activity [9]. 

The inheritance patterns also differ between FSHD1 and FSHD2. FSHD1 follows an autosomal dominant pattern, meaning an individual only needs to inherit one allele of chromosome 4 at region 4q35 that includes a contraction in the D4Z4 array to develop the disease, given that the contraction is located on a permissive 4qA allele [10]. On the other hand, FSHD2 follows a digenic pattern [11]. For an individual to inherit this type of FSHD, they need to have inherited an SMCHD1 mutation, as well as have a moderately sized D4Z4 array that is between approximately 8 and 20 repeat units, given that the array is located on a permissive 4qA allele [11].

Variation between patients can be attributed to numerous factors, such as D4Z4 array repeat unit quantity, D4Z4 methylation, and SMCHD1 mutation type [12]. Results from one study suggest that the clinical severity of FSHD1 in patients with 1 to 6 D4Z4 repeat units primarily depends on the D4Z4 array repeat unit quantity: within this range, patients with fewer repeat units displayed higher clinical severity [12]. Methylation levels could have less of an impact on clinical severity in these individuals because the length of the array is already shorter, in comparison to individuals with 7 to 10 D4Z4 repeat units. The study suggests that the clinical severity of FSHD1 in patients with 7 to 10 D4Z4 repeat units also depends on D4Z4 methylation, which aligns with this concept [12]. This same study found that the clinical severity experienced by FSHD2 patients within the sample depended on SMCHD1 mutation type in combination with D4Z4 repeat unit quantity [12]. Specifically, they found that individuals with a dominant negative mutation were more severely clinically affected than individuals with a haploinsufficiency mutation [12]. A dominant negative mutation leads to the production of a defective protein that hinders healthy SMCHD1 protein function, while a haploinsufficiency mutation results in insufficient SMCHD1 protein production [12].

#### Symptoms
Typically, an individual with FSHD can begin experiencing symptoms at any age due to the high variability of the clinical severity of this disease across patients. It is most common for patients with FSHD to begin experiencing symptoms in their teenage and adult years [14]. It is estimated that only approximately 5% of individuals with FSHD present symptoms before the age of 10 [14]. However, it is important to mention how this statistic has the potential to be higher in actuality. For instance, the unsymmetric nature of this disease can cause weakness in a certain region to appear as just an injury, rather than FSHD. 

FSHD affects multiple regions of the body because the skeletal muscles it degrades make up the face, upper body, and lower body. Symptoms most commonly appear in the face and upper body before the lower body. One study in particular that involved 34 patients with FSHD found that the first presentation of symptoms occurred in the facial muscles for 68.8% of the patients and occurred in the shoulder-girdle muscles for 28.1% of the patients [15]. 

The most common symptoms that appear involving the facial muscles include difficulty whistling and using a straw when drinking, eyes that do not close completely while sleeping, and even difficulty making certain facial expressions [14]. The most common symptoms that appear involving the upper body include protruding shoulder blades, known as “winging” and difficulty raising the arms above shoulder height [14]. The most common symptoms that appear involving the lower body include protruding stomachs, lower back curvature, as well as inability to lift the front foot when walking, causing an abnormal gait [14]. Overall muscle mass loss can occur in these regions as healthy muscle is replaced by fat and fibrous tissue due to FSHD [10].

#### Diagnosis
As mentioned previously, due to the nature of FSHD’s symptom presentation, the symptoms are often misattributed to other causes and age of diagnosis depends on clinical severity of symptoms, as in the case of Claire Anderson and her mother [13]. However, there are various methods used to diagnose FSHD, particularly genetic testing, since the disease is inherited. After a physical examination of the patient for the symptoms outlined previously, the physician may inquire about family history. Then, if the physician suspects a neuromuscular disorder, genetic testing can help determine which disease the patient has through differential diagnosis [17]. For instance, symptoms of FSHD, such as muscle mass loss and fat replacement, also appear in other diseases such as chronic sarcoid myopathy, an inflammatory muscle disease [16]. Specific genetic testing methods include Southern blotting, methylation analysis, and Nanopore long-read sequencing [17-19]. Southern blotting is used to measure the length of the D4Z4 fragment and to determine if the chromosome is 4qA [17]. Hence, it is primarily used to diagnose FSHD1 if the fragment is shorter than normal, about 1 to 10 repeat units, and on a 4qA chromosome [17]. Methylation analysis is used to display how methylated the D4Z4 region is [17]. Therefore, it is primarily used to diagnose FSHD2, in which the D4Z4 fragment is a typical length but atypically low in methylation [17]. Nanopore long-read sequencing is a newer technique that can evaluate the size of the D4Z4 fragment, whether it is on a 4qA chromosome, the type of gene variant in FSHD2, and methylation patterns [19]. As a result, this single method can be used to diagnose both FSHD1 and FSHD2 [19].

#### Standard of Care Treatments and Reimbursement
As of today, there are no approved treatments that help halt or slow the progression of FSHD1 or FSHD2 [17]. That being said, there are still approved treatments; they are just supportive ones, which involve the mitigation of symptoms, not the root cause. Hence, the supportive treatments that each patient may receive are dependent on the clinical severity of their symptoms, which is highly variable across patients [9].

After a patient is diagnosed with FSHD, multiple evaluations will help determine which supportive treatments would best serve the patient. For instance, a physical examination and a physical therapy evaluation will help assess the patient’s strength, as well as the limitations of their musculoskeletal system [17]. Based on these results, an individualized exercise plan can be formulated [17]. For example, the plan may include moderate aerobic exercise and strength training, which have been demonstrated to be safe in FSHD to some degree [17]. Additionally, mobility aids, such as wheelchairs, canes, and braces, can provide relief and help prevent falls if needed [17]. If a patient experiences difficulty raising their arms above shoulder height, depending on the severity, the patient may be able to get a surgical fixation of the scapula to the chest wall if they meet additional requirements [17]. Nonsteroidal anti-inflammatory drugs can be utilized to improve short-term pain, while antidepressants or anti-seizure medications can sometimes be utilized to improve chronic pain [17]. If a patient’s eyes do not close completely while they sleep, eye drops may help, and in more severe cases, using specialized tape to seal the eyelids shut during sleep may be necessary [17]. If a patient experiences difficulty breathing due to a weakened diaphragm or weakened abdominal muscles, they may require ventilatory support [17]. If a patient develops hearing loss, standard hearing loss treatments can be implemented, such as the usage of hearing aids [17].

Evidently, the symptoms of FSHD can lead to a variety of complications that vary from patient to patient, and these symptoms have supportive treatments; however, the underlying cause of FSHD cannot yet be treated, highlighting the importance of finding an FSHD-specific treatment, which can hopefully be accomplished through clinical trials soon.

It is also important to discuss the reimbursement of supportive treatments that are currently available. Insurance does not always completely cover the expenses that come along with the variety of doctor appointments, co-pays, medical equipment, caregiving, and necessary home modifications, as well as other expenses discussed in the Economic Burden section [20]. Recall how in that section, the study conducted in the United States found that the total out-of-pocket cost per year is over $28,000 per patient [6]. Additionally, another study found that FSHD patients with commercial insurance had significantly higher medical claims costs in comparison to individuals with the same type of commercial insurance who did not have FSHD [21]. These findings suggest that patients carry a substantial financial burden from FSHD, which also highlights that better insurance coverage for supportive treatment measures is a necessity.

#### Disease Progression & Prognosis
As mentioned previously, patients with FSHD most commonly begin experiencing symptoms in their teenage or early adult years [14]. Only approximately 5% of individuals with FSHD are estimated to show symptoms before the age of 10 [14]. Additionally, males with FSHD tend to show symptoms at a younger age than females [9]. Patients who have FSHD and present symptoms in early childhood are at a higher risk of faster disease progression and additional symptoms [9,17]. For instance, they are more susceptible to audiologic and ophthalmologic complications, such as hearing loss and retinal vasculopathy [17]. This could be, in part, due to early-onset FSHD being associated with very short D4Z4 fragments, and fewer repeat units are linked to more severe disease [12,15].

As discussed earlier, there is a descending pattern of symptom presentation that is most common among FSHD patients. More specifically, symptoms typically appear first in the face, then the shoulders and upper arms, and then the torso, legs, and feet [20]. This progression often occurs in an asymmetrical manner, meaning one side of the body may be more affected than the other [20]. The rate of progression varies but is typically slow and can span decades [20]. Although, the decline is not always gradual. Many patients notice certain muscles weakening quickly over a short period of time, and then their condition plateaus until the cycle repeats [21]. Importantly, even though FSHD worsens over time, it does not typically shorten a patient’s lifespan [21]. However, around 20% of people older than 50 years of age will rely on a wheelchair. 

Because FSHD symptoms and age of onset vary between patients, the disease progression and prognoses do as well. For instance, family members who share the same mutation can be affected very differently [20]. Surprisingly, this difference has even been observed in identical twins [9]. However, disease progression and prognosis are difficult to measure in studies, which can, in part, contribute to why there is not yet an approved FSHD treatment [10]. Complications in measuring disease progression and prognosis arise because the disease is asymmetrical, affects various muscles throughout the body, and typically progresses too slowly to be measured within the length of a typical clinical trial [10]. Additionally, human movement is extremely complex, and FSHD can affect it in different ways [10]. More specifically, FSHD can result in skeletal frame and muscle changes, as well as lead to compensatory movement mechanisms [10].

#### Biological Mechanisms: Anatomy, Organ Physiology, Cell and Molecular Physiology
Over time, DUX4 activity leads to damage at the molecular, cellular, tissue, organ, and anatomical levels [10,25].

At the molecular level, skeletal muscle force is driven by sarcomeres and the filament activity that occurs within them [25]. Sarcomeres consist of two different types of filaments, thin actin filaments and thick myosin II filaments, which are anchored at the Z discs and the M line, respectively [25,26]. When the muscle is not actively contracting, tropomyosin wraps around the actin filament and blocks the sites where myosin would bind, and it is held in place by the troponin complex, in which Troponin T binds tropomyosin and Troponin I binds actin [26]. When calcium is released from intracellular stores, it binds to the Troponin C subunit, which prompts a conformational change in Troponin C that prevents Troponin I from binding to actin. This unlocks tropomyosin from the myosin binding sites, allowing myosin to bind [26]. This allows for the commencement of the actin-myosin cross-bridge cycle, in which myosin uses energy from ATP hydrolysis to pull the actin filaments, and the Z discs, closer together, generating muscle force [25,26].

At the cellular level, sarcomeres are organized into myofibrils, and the myofibrils are organized into muscle fibers, which are muscle cells [25]. The disruption of cellular function due to the re-expression of DUX4 leads to muscle degeneration [9]. This degeneration causes the replacement of muscle fibers with fat and connective tissue, resulting in fewer muscle fibers with the contractile machinery that healthy muscle fibers have, and therefore decreased force generation [25].

At the organ level, the force generated by an entire muscle is dependent on the muscle quantity and quality, which are represented by lean muscle volume and fat percentage, respectively [25]. When the lean muscle volume decreases, so does the force generated [25]. When the fat percentage increases, the force generated decreases because the effective peak isometric stress decreases [25]. These relationships are explained through the joint torque equation, which estimates the rotational effect that muscles produce about a joint. The equation is approximately the sum, across all muscles crossing the joint, of each muscle’s lean volume divided by its optimal fiber length, multiplied by its peak isometric stress, multiplied by the perpendicular distance between the center of the joint and the muscle’s line of action [25]. Moreover, for the joint to be able to move and not remain in static equilibrium, the torque created by the muscles must be greater than the torque produced by the applied load. Hence, with regard to FSHD, a lower ability of the muscle to generate force will result in a lower torque created by the muscles, meaning that applied loads, such as the weight of the arm itself, become more difficult to overcome and intended movement to ensue. 

At the anatomical level, the replacement of healthy muscle fibers with fat and connective tissue, which leads to an inability to perform an intended movement, can lead to compensatory movements to still achieve the intended movement, which can lead to even more impairments [10]. It can also lead to changes in the skeletal frame, such as bone shape and posture, which can amplify compensatory mechanisms. The cycle repeats as muscle atrophy continues to ensue. 

Because there is currently no approved FSHD treatment that targets its underlying cause [17], this highlights the importance of developing one through research, hopefully in the near future. For instance, a therapy could target the prevention of the replacement of healthy fibers with fat and connective tissue through targeting DUX4 itself, as reducing its atypical re-expression could prevent damage at every level downstream [9]. Biomarkers, which are measurable characteristics that indicate biological or disease processes, as well as responses to a therapy, can be used to detect whether a therapy could be associated with intentional or unintentional changes within the body [10]. For example, specific biomarkers can be used if a patient were to be given a therapy that is intended to limit the replacement of healthy fibers with fat and connective tissue through targeting DUX4 itself. For instance, at the molecular and cellular levels, a muscle biopsy or blood sample could be extracted and tested for levels of KH Domain Containing 1-Like (KHDC1L), a biomarker that is indicative of DUX4 activity, as well as creatine kinase (ck), a biomarker that is indicative of muscle damage [27]. A clinical trial of an experimental therapy, delpacibart braxlosiran (del-brax), that was intended to target DUX4 activity utilized these exact biomarkers to evaluate the implications of the therapy on these levels to then evaluate the impact of the therapy on DUX4 activity so the clinical trial could move to the following phase [27]. At the tissue and organ levels, the prevention of fat replacement and the preservation of lean muscle volume could be evaluated through magnetic resonance imaging (MRI) [10,25]. More specifically, an MRI can produce a water image and a fat image. Then, a fat percentage image can be calculated by multiplying 100 by the fat signal in each pixel, divided by the sum of the fat and water signals in that same pixel [25].

#### Clinical Trials/Next-gen Therapies
There are numerous steps that need to ensue before an experimental FSHD therapy could be approved for usage in a non-experimental context: Preclinical, Phase 1, Phase 1/2, Phase 3, and FDA review [10]. In the case of FSHD clinical trials, during the Preclinical phase, the experimental therapy is evaluated in the context of the biological mechanisms of FSHD [10]. If successful, Phase 1 ensues, which evaluates if the therapy would be safe for humans to eventually use [10]. If successful, Phase 2 ensues, which involves selection of dosage, because this would vary drastically between, say, a mouse model and a human [10]. Additionally, this phase evaluates what early success would look like in clinical trials [10]. If successful, clinical trials comparing a therapy with a placebo can be conducted to test the effectiveness of the therapy using a pre-selected outcome measurement [10]. If successful, the therapy must be reviewed by the FDA [10].

Oftentimes, a therapy can appear promising during the earlier phases and still not succeed in Phase 3 [10]. For instance, losmapimod was developed to lower DUX4 expression by blocking a signaling pathway called p38 MAPK [10]. It was considered safe in Phase 1, and the MRI and functional results from its Phase 2 trial were considered promising [10]. However, in Phase 3, patients who received losmapimod did not improve meaningfully more than patients who received the placebo with regard to the functional results [10]. These results may support the suggestion in the Disease Progression and Prognosis section that functional results can be limited when evaluating FSHD due to reasons such as this disease resulting in skeletal frame and muscle changes, as well as leading to compensatory movement mechanisms, and the asymmetrical nature of the disease [10].

## Data Set:

#### The subject and visit/timepoint(s) analyzed: 
The dataset tracks a clinical cohort of 34 human subjects diagnosed with Facioscapulohumeral Muscular Dystrophy (FSHD). These participants were evaluated across three designated longitudinal timepoints. A baseline (Month 0), Year 1 (y1 and a mean follow-up of 13.45 months), and Year 2 (y2 and a mean follow-up of 27.88 months), enabling both cross-sectional and prospective assessments of disease progression.[1][10][25]

#### The tibialis anterior MRI data: 
Cross-sectional magnetic resonance imaging (MRI) was acquired axially through the lower extremities with a dedicated focus on the bilateral tibialis anterior (TA) muscles. The tibialis anterior was targeted because it is an early, prominent, and often asymmetrically affected muscle in FSHD that directly dictates functional ankle dorsiflexion and gait stability.[1][14][23][25]

#### The water-image and fat-image files: 
The MRI acquisition utilizes two-point chemical-shift Dixon imaging to yield co-registered, two-dimensional spatial arrays separated into distinct water-only and fat-only image volumes. Stored as 32-bit floating-point NumPy arrays (.npy) of matrix size 266 x 224 pixels, the fat images highlight subcutaneous and intramyocellular lipid signals, while the water images capture residual unreplaced muscle tissue and bulk water.[1][25]

#### The tibialis anterior segmentation mask: 
Anatomical segmentation masks are provided as unsigned 8-bit integer arrays (uint8) matching the 266 x 224 image grid. Each mask encodes a binary assignment ($1$ for muscle tissue, $0$ for background), delineating the cross-sectional perimeter of the tibialis anterior muscle within the anterior crural compartment while excluding adjacent osseous structures (tibia and fibula) and subcutaneous adipose tissue.[1][25]

#### The sampled locations along the muscle: 
The imaging data sample sequential, contiguous axial slices along the proximodistal length of the lower leg. This spatial sampling spans from the proximal origin of the tibialis anterior near the lateral tibial condyle and fibular head down toward the distal myotendinous junction above the ankle.[1][25]

#### The meaning and units of the image signals: 
The pixel values in the raw water and fat images represent transverse proton magnetization signal amplitudes recorded in arbitrary units (a.u.). When processed through chemical-shift Dixon decomposition, the calculated fat fraction (FF) yields a normalized, dimensionless percentage ranging from 0% to 100%, indicating the relative volumetric proportion of lipid protons to total proton density within each voxel.[1][25]

#### The clinical, biomarker, or functional measurements included in the dataset:
1. Genetic & Biomarker Measures: D4Z4 repeat array unit length and downstream DUX4 transcriptional signature expression scores assessed bilaterally. [1][11][12]   
2. Active Inflammation Biomarkers: Quantitative Short Tau Inversion Recovery (STIR) hyperintensity values to detect active muscle edema and inflammation.[1]   
3. Quantitative Strength Testing: Bilateral Quantitative Muscle Testing (QMT) measuring isometric ankle dorsiflexion force.[1][10]
4. Functional Mobility Assessments: Timed Up and Go (TUG), 10-meter walk/run (WR10), 4-step ascend, and 4-step descend times.[1][10]   
5. Composite Clinical Scoring: Ricci Clinical Severity Scores (CSS) scaled from 0 to 8.[1][23]

#### The source of the data: 
The clinical, biomarker, and MRI data originate from standardized multi-center FSHD natural history cohorts designed for clinical trial readiness (e.g., protocols documented by Statland et al., 2018, and Wang et al., 2021). Axial Dixon acquisitions were conducted on 3.0T MRI scanners utilizing phased-array lower-extremity surface coils. Binary TA masks were produced through manual or semi-automated boundary segmentation by trained imaging readers using anatomical fasciae (tibia, interosseous membrane, and anterior intermuscular septum) as delineating boundaries. [1][10][25]

#### Explain how the MRI data and muscle mask were generated or provided for this project. Cite the paper, dataset documentation, or other sources that informed your description:

The MRI datasets and associated muscle segmentation masks were generated and provided as part of clinical trial readiness natural history protocols for Facioscapulohumeral Muscular Dystrophy (FSHD). Quantitative chemical shift Dixon imaging was conducted on clinical 3.0-Tesla MRI scanners using phased array lower extremity surface coils. Two-point Dixon gradient-echo sequences acquired in-phase and out-of-phase echoes at slightly offset echo times (T_E), exploiting the precessional frequency difference between water and lipid protons. Vendor image reconstruction algorithms separated these multi-echo acquisitions into co-registered, quantitative water-only and fat-only volumetric series. The binary segmentation masks for the tibialis anterior muscle were generated through expert anatomical contouring on contiguous axial slices. Trained imaging analysts outlined the muscle boundaries using anatomical landmarks—specifically confining the region of interest within the anterior crural compartment bounded by the lateral tibial cortex, fibular shaft, crural fascia, and anterior intermuscular septum, while excluding surrounding subcutaneous fat and cortical bone to prevent partial volume artifacts. [10][25]

## MRI Data Analysis Methdology:

#### How the water, fat, and segmentation-mask files were loaded: 

Image files and anatomical masks are loaded into python using the numpy library specifically the numpy.load() function, which imports the data as 2D numerical arrays [25]

#### How the tibialis anterior mask was used to identify the muscle:

The binary mask array (mask_img) contains values of 1 inside the tibialis anterior and 0 elsewhere. By converting this mask into a boolean condition (mask_img > 0), it serves as an indexing filter applied directly to the 2D fat fraction map. This isolates the target muscle voxels while excluding all non-target structures, including the tibia, fibula, deep posterior compartments, and the subcutaneous adipose layer. [1][25]

#### How fat fraction was calculated for the whole image region and for the TA muscle only:

The two-point Dixon fat fraction (FF) map is computed pixel-by-pixel across the full field-of-view by taking the ratio of fat signal intensity to the sum of fat and water signal intensities:

FF_pixel = (S_fat/(S_fat + S_water)) x 100% 

To prevent division-by-zero artifacts in background air regions where both signals approach zero, conditional masking is applied. The muscle-specific fat fraction is subsequently obtained by extracting only the voxels where the binary mask evaluates to true [1][25]

#### How the fat-fraction distribution was summarized: 

The distribution of fat fractions within the tibialis anterior ROI was visualized using a binned frequency histogram with overlay lines for the parametric mean (58.50%) and non-parametric median (61.82%). The broad distribution spans from preserved low-fat muscle voxels (<20%) to near-complete fatty replacement (>80%), capturing the heterogeneous, mosaic pattern of fibro-adipose substitution characteristic of FSHD. [1][23][25]

#### Which summary statistics were calculated, such as mean, median, standard deviation, and number of muscle pixels:

Four primary quantitative summary metrics were computed from the isolated muscle voxel array:
1. Number of Muscle Pixelsv(N): Represents the cross-sectional area of the muscle slice.
2. Mean Fat Fraction (mu): The arithmetic average of all voxel fat fractions in the ROI
3. Median Fat Fraction (x(tilde)): The central value of the distribution, providing resistance against peripheral partial volume effects.
4. Standard Deviation (sigma): Reflects intramuscular fat infiltration heterogeneity.

#### How measurements were summarized across locations along the muscle:
Across the contiguous axial stack, slice-specific muscle statistics are combined to compute a composite, volume-weighted mean fat fraction:

FF_vol = Sum from slice i = 1 to M (FF_i x Ni)/ sum from slice i = 1 to M (N_i)\

where FF_iand N_i denote the mean fat fraction and muscle pixel count of slice i, respectively. In addition, the spatial variance of fat replacement along the proximodistal axis (fvar) is calculated to quantify longitudinal gradient differences between proximal muscle bellies and distal myotendinous junctions. [1][25] 

#### If you compare visits or use interpolation in your final analysis, describe those methods clearly, including the variables used and the assumptions made:

To compare longitudinal scans across visits (Baseline, Year 1, Year 2), differences in patient positioning and slice prescription require spatial normalization. Slice locations along the leg are mapped to a normalized anatomical coordinate system ranging from $0\%$ (distal myotendinous junction) to $100\%$ (proximal fibular head). One-dimensional linear spline interpolation is performed across this normalized spatial coordinate to resample muscle fat fraction and cross-sectional area profiles to matching anatomical intervals across timepoints. This assumes that longitudinal muscle geometry changes monotonically and that slice spacing remains uniform between acquisitions. [10][25]

## Additional Analysis and Conclusions:

To address the project goal, mean change and responsiveness were evaluated across an interpolated interval of 12 to 22 months for each measure: the progression of muscle tissue through MRI fat fraction in both legs, force output through dorsiflexion strength in both legs, task performance through the 10 m Walk/Run, and daily impact through the Clinical Severity Score [1].

```python
# Code Used for Displaying Interpolation and Responsiveness of FSHD Measures Between 12 and 22 Months

import pandas as pd # for data handling
import numpy as np # for numerical calculations
import matplotlib.pyplot as plt # for plotting

# Load the dataset 
DATA_PATH = "/Users/julialundy/Downloads/Computational/Module 1/BME_2315_Module2/Dataset/Module2_FSHD_Dataset.csv"   
df = pd.read_csv(DATA_PATH)

# Months since baseline for each visit (baseline, year-1, year-2) are already in the dataset as lag_months_baseline, lag_months_y1, lag_months_y2. We'll use these to interpolate each biomarker at two time points of interest.
TIME_1, TIME_2 = 12, 22

# Interpolation will be done for the following biomarkers, with the specified columns for each visit, units, and direction that indicates worsening:
V = ["baseline", "y1", "y2"] 
BIOMARKERS = {   # name: (columns at baseline/y1/y2, units, direction that means "worse")
    "TA Fat Fraction (Right)":     ([f"ff_{v}_R" for v in V], "% fat", "up"), 
    "TA Fat Fraction (Left)":     ([f"ff_{v}_L" for v in V], "% fat", "up"),
    "Dorsiflexion (Right)":        ([f"dorsiflexion_qmt_R_{v}" for v in V], "QMT force", "down"),
    "Dorsiflexion (Left)":        ([f"dorsiflexion_qmt_L_{v}" for v in V], "QMT force", "down"),
    "10 m Walk/Run":           ([f"wr_10_{v}" for v in V], "seconds", "up"),
    "Clinical Severity Score": ([f"clinical_severity_score_{v}" for v in V], "score", "up"),
}

# interpolate a biomarker at a given time point for one patient, using the measured values at the three visits
def interp_at(row, cols, t):
    """Estimate a biomarker at time t (months) from one patient's measured visits."""
    pts = pd.DataFrame({"t": [row[f"lag_months_{v}"] for v in V],
                        "y": [row[c] for c in cols]}).dropna()
    pts = pts.drop_duplicates(subset="t").sort_values("t")   # e.g. 1-0031 has two visits at 24.3 mo
    if len(pts) < 2 or t < pts["t"].min() or t > pts["t"].max():
        return np.nan                                        # outside measured range -> don't guess
    return np.interp(t, pts["t"], pts["y"])
records = []
for _, r in df.iterrows():
    for name, (cols, units, worse) in BIOMARKERS.items():
        v1, v2 = interp_at(r, cols, TIME_1), interp_at(r, cols, TIME_2)
        records.append({"subject_id": r["subject_id"], "biomarker": name,
                        f"at_{TIME_1}mo": v1, f"at_{TIME_2}mo": v2, "change": v2 - v1})
long_df = pd.DataFrame(records)

# Summary statistics for each biomarker: n, mean change, SD, SEM, 95% CI, units, direction of worsening, and SRM (standardized response mean)
summary = (long_df.dropna(subset=["change"]).groupby("biomarker", sort=False)["change"]
           .agg(n="count", mean_change="mean", sd="std"))
summary["sem"] = summary["sd"] / np.sqrt(summary["n"])
summary["ci95"] = 1.96 * summary["sem"]
summary["units"] = [BIOMARKERS[b][1] for b in summary.index]
summary["worse_if"] = [BIOMARKERS[b][2] for b in summary.index]

# SRM signed so positive = worsening for every biomarker
summary["SRM_worsening"] = [m / s * (1 if BIOMARKERS[b][2] == "up" else -1)
                            for b, m, s in zip(summary.index, summary["mean_change"], summary["sd"])]

# Bar graph of mean change with 95% CI for each biomarker
fig, axes = plt.subplots(2, 3, figsize=(11, 6.5))
for ax, (name, s) in zip(axes.flat, summary.iterrows()):
    ax.bar([0], [s["mean_change"]], width=0.5, color="#3b6ea8",
           yerr=[s["ci95"]], capsize=6, error_kw={"lw": 1.2, "ecolor": "0.25"})
    ax.axhline(0, color="0.3", lw=0.8)
    lim = max(abs(s["mean_change"]) + s["ci95"], 1e-6) * 1.35
    ax.set_ylim(-lim, lim)
    ax.set_xlim(-0.8, 0.8)
    ax.set_xticks([])
    ax.set_title(name, fontsize=11)
    ax.set_ylabel(f"Mean Change ({s['units']})", fontsize=9)
    arrow = "↑ = worse" if s["worse_if"] == "up" else "↓ = worse"
    ax.text(0.97, 0.95, arrow, transform=ax.transAxes, ha="right", va="top", fontsize=9, color="0.35")
    ax.text(0.03, 0.05, f"n = {int(s['n'])}\nmean = {s['mean_change']:+.2f}",
            transform=ax.transAxes, fontsize=9, color="0.2")
    ax.spines[["top", "right", "bottom"]].set_visible(False)
    ax.grid(axis="y", color="0.92")
    ax.set_axisbelow(True)
fig.suptitle(f"Mean Change From {TIME_1} to {TIME_2} Months (Interpolated), ±95% CI", fontsize=13)
plt.tight_layout()
plt.show()

# Bar graph of SRM (standardized response mean) for each biomarker, signed so positive = worsening
srm = summary["SRM_worsening"].sort_values()
fig, ax = plt.subplots(figsize=(7.5, 4))
ax.barh(srm.index, srm.values, color="#3b6ea8", height=0.55)
for y, v in enumerate(srm.values):
    ax.text(v + (0.02 if v >= 0 else -0.02), y, f"{v:.2f}", va="center",
            ha="left" if v >= 0 else "right", fontsize=9)
ax.axvline(0, color="0.3", lw=0.8)
ax.set_xlim(min(srm.min(), 0) - 0.2, max(srm.max(), 0.8) + 0.1)
ax.set_ylim(-0.6, len(srm) - 0.5)
for x in [0.2, 0.5, 0.8]:   # dashed reference lines at small, moderate, and large SRM thresholds
    ax.axvline(x, color="0.8", ls="--", lw=0.8)
ax.set_xlabel("Standardized Response Mean (Positive = Worsening)", labelpad=8)
ax.set_title(f"Responsiveness, {TIME_1} to {TIME_2} Months")
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()

# Plot measured values and interpolated estimates for a few example patients
patients = ["1-0021", "1-0025", "13-0008"]
colors = ["tab:blue", "tab:orange", "tab:green"]

fig, axes = plt.subplots(2, 3, figsize=(13, 7))
for ax, (name, (cols, units, worse)) in zip(axes.flat, BIOMARKERS.items()):
    for pid, color in zip(patients, colors):
        r = df[df["subject_id"] == pid].iloc[0]
        measured = pd.Series([r[c] for c in cols], index=[r[f"lag_months_{v}"] for v in V]).dropna()
        ax.plot(measured.index, measured.values, "-o", color=color, label=f"Patient {pid}")
        estimates = [interp_at(r, cols, TIME_1), interp_at(r, cols, TIME_2)]
        ax.plot([TIME_1, TIME_2], estimates, "D", mfc="white", mec=color, mew=2, ms=9)
    ax.set_title(name)
    ax.set_xlabel("Months Since Baseline")
    ax.set_ylabel(units)

axes[0, 0].plot([], [], "D", mfc="white", mec="gray", mew=2, label="Interpolated Estimate")
fig.legend(*axes[0, 0].get_legend_handles_labels(), loc="lower center", ncol=4)
fig.suptitle("Measured Values (Circles) and Interpolated Estimates (Diamonds)")
plt.tight_layout(rect=[0, 0.06, 1, 1])
plt.show()

# Wilcoxon signed-rank test to test if the change from 12 to 22 months is significantly different from zero
from scipy.stats import wilcoxon

rows = []
for name, (cols, units, worse) in BIOMARKERS.items():
    change = np.array([interp_at(r, cols, TIME_2) - interp_at(r, cols, TIME_1) for _, r in df.iterrows()])
    change = change[~np.isnan(change)]
    p = wilcoxon(change).pvalue
    rows.append((name, str(len(change)), f"{p:.3f}"))

line = "+" + "-" * 27 + "+" + "-" * 6 + "+" + "-" * 9 + "+"
print(f"Change From {TIME_1} to {TIME_2} Months")
print(line)
print(f"| {'Biomarker':<25} | {'n':<4} | {'p-value':<7} |")
print(line)
for name, n, p in rows:
    print(f"| {name:<25} | {n:<4} | {p:<7} |")
print(line)
```

<b>Interpolation and Responsiveness of FSHD Measures Between 12 and 22 Months Analysis:</b> 

The MRI, biomarker, or function measurement that changed most consistently between the interpolated time frame of 12 and 22 months was the TA Fat Fraction. This may reflect the asymmetrical nature of FSHD. Importantly, none of the changes for any of the biomarkers were statistically significant when a Wilcoxon Signed-Rank Test was performed. For the TA Fat Fraction (Right) and TA Fat Fraction (Left), the p-values were 0.063 and 0.437, respectively, in the “Change from 12 to 22 Months” table; hence, both changes could not be distinguished from random variation around zero. Along with the TA Fat Fraction mean changes, the Dorsiflexion (Right) measurement at 12 and 22 months appeared to have a mean change of +0.79 QMT force units, as displayed in the “Mean Change From 12 to 22 Months (Interpolated), ±95% CI” bar graphs. The Dorsiflexion (Left) measurement at 12 and 22 months appeared to have a mean change of +0.21 QMT force units, which is evidently not as substantial as the Right. This may reflect the asymmetrical nature of FSHD. Interestingly, the mean change increase in QMT force suggests that dorsiflexion strength slightly improved in both legs, on average, within that time period. However, we would actually expect the opposite to occur due to FSHD progression leading to replacement of healthy muscle tissue with fat and connective tissue, hindering muscle strength and force output. This could have occurred due to measurement variability between visits, such as differences in patient effort or positioning, compensatory movement mechanisms, and other factors discussed in the Disease Progression and Prognosis section of the Disease Background section [10]. The Wilcoxon Signed-Rank Test resulted in p-values of 0.345 and 0.580 for the Dorsiflexion (Right) and the Dorsiflexion (Left) changes, respectively, in the “Change from 12 to 22 Months” table; hence, both changes could not be distinguished from random variation around zero. The mean change in seconds of the 10 m Walk/Run, as well as the mean change in Clinical Severity Score, were the least sensitive to differences over the time period comparatively, with mean changes of −0.03 seconds and +0.08 points, respectively, and p-values of 0.804 and 0.683, as displayed in the “Mean Change From 12 to 22 Months (Interpolated), ±95% CI” bar graphs and “Change from 12 to 22 Months” table.

Of the measures analyzed, the TA Fat Fraction shows the most promise as a biomarker and exploratory endpoint for a Phase 1/2 clinical trial in comparison to Dorsiflexion, the 10 m Walk/Run test, and the Clinical Severity Score. This measurement has fewer confounding variables in the data collection process and is not dependent on patient performance on any given day. As discussed previously, the Standardized Response Mean values of 0.30 and 0.20 for the TA Fat Fraction (Right) and TA Fat Fraction (Left), respectively, in the “Responsiveness, 12 to 22 Months” bar graph, are the largest (most positive) of all the measures and in the direction expected for disease progression. The Dorsiflexion (Right) value of −0.27 was not expected and does not align with disease progression, which hinders its promise as a biomarker in this case. Additionally, the Dorsiflexion (Left), 10 m Walk/Run, and Clinical Severity Score displayed Standardized Response Means closest to zero, −0.08, −0.03, and 0.08, respectively, and the latter measure is the only one of the three that aligns with what would be expected. However, since the TA Fat Fraction’s changes were not statistically significant, it would likely be most useful as an exploratory endpoint in a longer or larger clinical trial with standardized visit timing, which would also limit the need for interpolation. The irregular visit timing that required interpolation is displayed in the “Measured Values (Circles) and Interpolated Estimates (Diamonds)” plots, which show the same three patients for each biomarker from 0 to approximately 25 months.

Before evaluating treatment effectiveness, it would be necessary to obtain statistically significant results when evaluating responsiveness to determine which of the biomarkers would be best for measuring FSHD progression. As of right now, the most promising one appears to be the TA Fat Fraction biomarker; however, as explored above, there was no statistical significance to support this suggestion. Hence, another study could replicate this one, but include more patients, as well as only include patients who can commit to a standardized visit schedule before the commencement of the study to limit the need for interpolation, which could have hindered the validity of the results. However, as per the “Measured Values (Circles) and Interpolated Estimates (Diamonds)” plots, interpolation likely provides a reasonable estimate, especially when the visit where the actual measurement occurred was close to 12 or 22 months. Additionally, it would be a good idea to compare the responsiveness between patients with high, moderate, and low baseline TA Fat Fraction, because once a patient has a substantial TA Fat Fraction, such as Patient 13-0008, there are fewer muscle fibers left that can be replaced by fat in comparison to a patient with a lower TA Fat Fraction baseline, such as Patients 1-0021 and 1-0025. Although the TA Fat Fraction appears to be the most promising biomarker to utilize, it is still important to use secondary endpoints that account for the force output, task performance, as well as daily impact that are represented through Dorsiflexion strength testing, the 10 m Walk/Run, and the Clinical Severity Score, respectively. In terms of accessibility, MRI and MRI analysis can be costly and time-consuming. Additionally, this could result in unintended negative effects of the treatment occurring and going unnoticed. For instance, if the only measure being used in a clinical trial were the Clinical Severity Score, then the progression of muscle tissue, force output, and task performance would be neglected and could potentially decline depending on the side effects of the treatment. Therefore, when evaluating the effectiveness of a treatment before it is made available to patients, it is important to monitor for unintended side effects to determine whether the intended benefits outweigh them and whether the treatment is safe enough for widespread use.
